## Introduction:
The final output convolution layers of a trained image classification network contain detailed semantic information about the input image. These feature maps, although low resolution (e.g. 7 x 7) also contain spatial information about the location of the primary object in the image. In the **class activation mapping** technique we wish to extract this spatial information pertaining to the identified class and overlay this activation on the input image as shown in the figure below.

```

imagenette_dict={0:'Tench Fish',217:'English Springer',482:'Casette Player',
                491:'Chain Saw',497:'Church',566:'French Horn',
                 569:'Grabage Truck',571:'Gas Pump',574:'Golfball',
                 662:'Modem', 701:'Parachute',864:'Tow Truck', 867:'Trailer Truck'}
mean = [0.485, 0.456, 0.406]  # Example mean used for normalization
std = [0.229, 0.224, 0.225]   # Example std used for normalization

def unnormalize_image(image, mean, std):
    """
    Unnormalize a PyTorch image tensor.

    Args:
        image (torch.Tensor): The normalized image tensor of shape [C, H, W].
        mean (list or tuple): The mean used for normalization.
        std (list or tuple): The standard deviation used for normalization.

    Returns:
        torch.Tensor: The unnormalized image tensor.
    """
    # Remove the batch dimension
    image = image.squeeze(0)

    # Check if the image is in the shape [C, H, W]
    if image.dim() == 3:
        # Unnormalize the image
        for t, m, s in zip(image, mean, std):
            t.mul_(s).add_(m)
    else:
        raise ValueError("Expected image tensor of shape [C, H, W]")

    # Convert from [C, x, y] to [x, y, C]
    image = image.permute(1, 2, 0)

    return image
```

# Class Activation Mapping
You can implement the class activation mapping algorithm in the code cell below.



```
import matplotlib.pyplot as plt
import numpy as np

# Initialize Figure
fig = plt.figure(figsize=(10,10), frameon=False)

# Ensure model is eval mode to disable batch normalisation during inference.
model.eval()

dataiter = iter(data_loader)

# Iterate over the data_loader to produce a few examples
for i, (image, true_label) in enumerate(data_loader):
    if i >= 16:  # Limit to 16 examples for display
        break
    # Get a batch of data
    image, target_class = next(dataiter)

    # Store the original normalized image for unnormalizing later for display
    original_normalized_image_for_display = image.clone()

    #image = image.to(device)

    # Pass data through model
    outputs = model(image)
    op_features = outputs['op_features']
    op_logits = outputs['op_logits']

    # Get the predicted class
    #predicted_class = torch.argmax(op_logits, dim=1)

    # Get predicted class
    probabilities = F.softmax(op_logits, dim=1)
    max_prob, predicted_class = torch.max(probabilities, 1)

    # Get the weights of the classifier layer
    # The state_dict was obtained in the previous cell
    classifier_weights = sd['fc.weight'][predicted_class.item()]

    # Reshape feature map to (C, H*W)
    feature_map = op_features.squeeze(0) # Remove batch dimension: [1, C, H, W] -> [C, H, W]

    # Calculate the Class Activation Map
    # CAM = sum(weight_c * feature_map_c)
    cam = torch.einsum('cwh,c->wh', feature_map, classifier_weights)

    # Apply ReLU to the CAM (as done in the original paper)
    cam = F.relu(cam)

    # Normalize the CAM to values between 0 and 1
    cam_min, cam_max = cam.min(), cam.max()
    act_map = (cam - cam_min) / (cam_max - cam_min + 1e-5) # Add small epsilon to avoid division by zero

    # Upsample the CAM to the original image size
    act_map_us = F.interpolate(act_map.unsqueeze(0).unsqueeze(0), size=(HW_trg, HW_trg), mode='bilinear', align_corners=False).squeeze()

    # N. Plotting
    def get_value(dictionary, key):
        try:
          return dictionary[key]
        except KeyError:
           #return "Error"
           return str(key)
    #label = get_value(imagenette_dict,int(predicted_class.cpu().numpy())) # Added .cpu() here
    label = get_value(imagenette_dict,int(predicted_class.numpy()))

    ax = fig.add_subplot(4, 4, i + 1, xticks=[], yticks=[])
    ax.title.set_text(label)
    img_to_display = unnormalize_image(original_normalized_image_for_display, mean, std)
    ax.imshow(img_to_display)
    ax.imshow(act_map_us.cpu().detach().numpy(), cmap=plt.cm.jet, alpha=0.4)

plt.tight_layout()
plt.show()
```
